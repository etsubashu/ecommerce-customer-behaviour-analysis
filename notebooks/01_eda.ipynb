5. Univariate Analysis

Gender Distribution

```python
plt.figure(figsize=(7,5))

sns.countplot(x=df["Gender"])

plt.title("Gender Distribution")

plt.show()
```

This bar chart visualizes the distribution of customers across different gender categories in the dataset. The graph shows that the Female and Male categories have the highest number of records, while other gender categories such as Bigender, Genderfluid, Non-binary, Polygender, Agender, and Genderqueer have smaller representations. This indicates that the dataset is primarily composed of male and female customers, with limited samples from other gender groups.

Age Distribution

```python
plt.figure(figsize=(8,5))

sns.histplot(df["Age"], kde=True)

plt.title("Age Distribution")

plt.show()
```

This histogram visualizes the distribution of customer ages in the dataset. The bars represent the frequency of customers within different age ranges, while the density curve shows the overall distribution pattern. The graph indicates that customer ages are relatively evenly distributed between approximately 18 and 50 years old, suggesting that the dataset includes a diverse range of age groups for analyzing customer purchasing behavior.

Income Level Distribution

```python
plt.figure(figsize=(8,5))

sns.countplot(x=df["Income_Level"])

plt.title("Income Level Distribution")

plt.show()
```

This bar chart visualizes the distribution of customers across different income levels in the dataset. The graph compares the number of customers categorized as Middle and High income levels. The results show that both categories have a relatively balanced distribution, with the High income group having a slightly larger number of customers. This indicates that the dataset contains diverse customer income profiles for analyzing purchasing behavior.

Correlation Heatmap

```python
import numpy as np

numeric_df = df.select_dtypes(include=np.number)

plt.figure(figsize=(12,8))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()
```

This correlation heatmap visualizes the relationships between numerical variables in the dataset. Each cell represents the correlation coefficient between two variables, with values ranging from -1 to 1. Positive values indicate a positive relationship, while negative values indicate an inverse relationship. The heatmap shows that most variables have weak correlations with each other, as the values are close to zero. This suggests that the features are relatively independent and there is little multicollinearity in the dataset.

Purchase Amount by Gender

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    x=df["Gender"],
    y=df["Purchase_Amount"]
)

plt.title("Purchase Amount by Gender")

plt.show()
```

This boxplot visualizes the distribution of Purchase_Amount across different gender categories. The graph compares customer spending behavior by showing the median, spread, and variability of purchase amounts for each gender group. The boxes represent the interquartile range (IQR), while the whiskers indicate the range of non-outlier values. The visualization suggests that purchase behavior varies across gender categories, with some groups showing higher variability in spending patterns than others.

Income Level vs Purchase Amount

```python
# Convert Purchase_Amount to numeric first

df["Purchase_Amount"] = df["Purchase_Amount"].replace('[\$,]', '', regex=True)

df["Purchase_Amount"] = df["Purchase_Amount"].astype(float)

# Scatter Plot

plt.figure(figsize=(8,5))

sns.scatterplot(
    x=df["Income_Level"],
    y=df["Purchase_Amount"]
)

plt.title("Income Level vs Purchase Amount")

plt.xlabel("Income Level")
plt.ylabel("Purchase Amount")

plt.show()
```

This scatter plot visualizes the relationship between Income_Level and Purchase_Amount. The graph compares customer purchase amounts across different income groups, specifically Middle and High income levels. The distribution of points shows that customers from both income categories have a wide range of purchase amounts, indicating that purchase behavior varies among customers regardless of income level. The visualization helps analyze customer spending patterns and identify possible trends between income and purchasing behavior.

Pairplot

```python
sns.pairplot(
    df[
        [
            "Age",
            "Purchase_Amount",
            "Product_Rating",
            "Time_to_Decision"
        ]
    ]
)

plt.show()
```

Outlier Detection

Purchase Amount Boxplot

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    x=df["Purchase_Amount"]
)

plt.title("Outlier Detection for Purchase Amount")

plt.xlabel("Purchase Amount")

plt.show()
```

This boxplot visualizes the distribution of the Purchase_Amount variable and helps detect potential outliers in the dataset. The central line inside the box represents the median purchase amount, while the box represents the interquartile range (IQR), showing the spread of the middle 50% of the data. The whiskers indicate the minimum and maximum values within the acceptable range. Since no extreme points appear outside the whiskers, the dataset shows little or no significant outliers after preprocessing.
