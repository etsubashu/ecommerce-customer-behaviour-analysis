1. Handling Missing Values

Missing values are handled to make the dataset complete. Numerical missing values are filled using the median, while categorical missing values are filled using the mode.

```python
df = pd.read_csv("../data/raw data.csv")
```

```python
df.isnull().sum()
```

```python
# Fill categorical missing values

df["Social_Media_Influence"] = df["Social_Media_Influence"].fillna("Unknown")
```

```python
df["Engagement_with_Ads"] = df["Engagement_with_Ads"].fillna(
    df["Engagement_with_Ads"].mode()[0]
)
```

This step handles missing values in categorical columns. Missing values in the Social_Media_Influence column are replaced with the category "Unknown", while missing values in the Engagement_with_Ads column are replaced using the most frequent value (mode). This ensures the dataset remains complete without losing important information.

2. Handling Inconsistencies

```python
df["Purchase_Amount"] = df["Purchase_Amount"].replace(
    '[\$,]',
    '',
    regex=True
)
```

This step removes special symbols such as dollar signs ($) and commas (,) from the Purchase_Amount column. These symbols prevent numerical analysis and machine learning processing. After removing the symbols, the column becomes suitable for conversion into a numerical data type.

```python
# Convert to float

df["Purchase_Amount"] = df["Purchase_Amount"].astype(float)
```

After removing special symbols from the Purchase_Amount column, the data type is converted into float format. This allows numerical calculations, statistical analysis, visualization, and machine learning operations to be performed correctly.


```python
print(df["Purchase_Amount"].dtype)
```

This step checks the data type of the Purchase_Amount column after conversion. It verifies that the column has been successfully converted into a numerical float data type for further analysis and machine learning processing.

3. Duplicate Removal

```python
# Check duplicates

df.duplicated().sum()
```

This step checks the dataset for duplicate records using the duplicated() function. Duplicate rows can negatively affect data quality and machine learning performance, so identifying them is important before model training.

```python
# Remove duplicates

df = df.drop_duplicates()
```

This step removes duplicate records from the dataset using the drop_duplicates() function. Removing duplicate rows improves data quality and prevents repeated information from affecting machine learning model performance.

4. Outlier Treatment

```python
Q1 = df["Purchase_Amount"].quantile(0.25)

Q3 = df["Purchase_Amount"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR

upper = Q3 + 1.5 * IQR

df = df[
    (df["Purchase_Amount"] >= lower) &
    (df["Purchase_Amount"] <= upper)
]
```

This step detects and removes outliers from the Purchase_Amount column using the Interquartile Range (IQR) method. The first quartile (Q1), third quartile (Q3), and IQR are calculated to determine acceptable value ranges. Values outside the lower and upper bounds are treated as outliers and removed from the dataset to improve data quality and model performance.

5. Feature Engineering

```python
df["Spending_Per_Age"] = (
    df["Purchase_Amount"] / df["Age"]
)
```

This step creates a new feature called Spending_Per_Age through feature engineering. The feature is calculated by dividing Purchase_Amount by Age, which helps represent customer spending behavior relative to age. Feature engineering can improve machine learning model performance by providing additional useful information.

```python
from sklearn.preprocessing import LabelEncoder, StandardScaler

# Remove ID-like columns if they exist
drop_cols = ["Customer_ID", "Phone_Number", "Phone", "Email", "Name"]

for col in drop_cols:
    if col in df.columns:
        df = df.drop(col, axis=1)

# Encode remaining categorical columns
le = LabelEncoder()

for col in df.select_dtypes(include="object").columns:
    df[col] = le.fit_transform(df[col].astype(str))

# Define X and y
X = df.drop("Purchase_Intent", axis=1)
y = df["Purchase_Intent"]

# Scale features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

This step prepares the dataset for machine learning model training. First, unnecessary ID-like columns such as customer IDs, phone numbers, emails, and names are removed because they do not contribute meaningful information to prediction. Categorical variables are then converted into numerical values using Label Encoding. After encoding, the dataset is divided into independent features (X) and the target variable (y). Finally, StandardScaler is applied to normalize the feature values, ensuring that all variables are on a similar scale and improving machine learning model performance.
